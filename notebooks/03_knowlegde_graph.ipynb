# Knowledge Graph: entity & relationship extraction

Flow: **load chunks → schema → prompt → extract one chunk → normalize names → validate → test on 15 chunks → review**

What changed after the first test (5 AGI chunks, only ~3 of 19 relations were correct):

| Problem in v1 | Fix here |
|---|---|
| `IS_A` / `RELATED_TO` / `INTRODUCES` used for everything ("GPT-4 IS_A mathematics") | Removed the vague types. Added `HAS_CAPABILITY`, `OUTPERFORMS`, `DEVELOPED_BY`, `EVALUATED_ON`… each with a definition and direction |
| Fields typed as DATASET, authors typed as ORGANIZATION | New `DOMAIN` type. Front matter, acknowledgments and author lists are skipped |
| "Artificial intelligence" vs "artificial intelligence", LLM vs large language model | `aliases` field + one alias map + case-insensitive names |
| Correct relations rejected because of `GPT--4` or curly quotes | Quotes and dashes are normalized before matching |
| Wrong relations accepted because the quote existed somewhere | **Both entity names must appear inside the quote**, type rules (e.g. `DEVELOPED_BY` must point to an ORGANIZATION), one quote can't back more than 3 relations |
| Test only covered 5 AGI chunks | Test set = 3 chunks from **each** of the 5 papers |

**v3 changes** (after the v2 test: 6 of 14 relations correct, 43%):

| Problem in v2 | Fix in v3 |
|---|---|
| Citations became entities ("Mei et al.", "Devlin et al. 2019") | Entities with `et al.` or `[...]` are dropped, and the prompt says so |
| Good relations rejected: the quote skipped "GPT-4" or dropped `[362]` | Citation markers removed before matching; names are checked in the **whole sentence** around the quote |
| "max 3 relations per quote" killed correct ones (GPT-4 IS_A LLM) | Rule removed. Instead: only one relation per entity pair per chunk |
| HELM "DEVELOPED_BY" language models, misuse risks called limitations | New `EVALUATES` and `HAS_RISK` types; `PART_OF` tightened; ORGANIZATION only allowed in `DEVELOPED_BY` |
| "examples include BERT, GPT-3, CLIP" gave nothing | New prompt example for that pattern |

**v3 test: 12 of 17 correct (71%), 16 of 17 with partials (94%).** After it, two last fixes:
the sentence check no longer runs into the next sentence, and a new `COVERS` relation (MMLU COVERS mathematics).

**Part 3** (after the full run): fixes that need no re-extraction, saved as a **v2** graph next to v1.

## 2. Schema: what the model is allowed to return

**Entity types**: what kind of thing a name is.
**Relation types**: each has a fixed **direction** (`source → target`) and a one-line meaning. The meaning goes into the prompt.
There's deliberately **no "RELATED_TO"**: if none of these fits, the model should return nothing.

```python
ENTITY_TYPES = {
    "MODEL":        "a specific AI model or system, e.g. GPT-4, PaLM, BERT",
    "ORGANIZATION": "a company, lab or university, e.g. OpenAI, Google, Stanford",
    "BENCHMARK":    "an evaluation suite or test, e.g. HELM, MMLU, HumanEval",
    "DATASET":      "a collection of data used for training, e.g. Common Crawl, The Pile",
    "METHOD":       "a technique or algorithm, e.g. RLHF, chain-of-thought prompting, transformer",
    "METRIC":       "something measured, e.g. accuracy, calibration, toxicity",
    "TASK":         "a concrete task, e.g. question answering, summarization",
    "DOMAIN":       "a field or subject area, e.g. medicine, law, mathematics, coding",
    "CONCEPT":      "an abstract idea, e.g. artificial general intelligence, homogenization, hallucination",
}

RELATION_TYPES = {
    "DEVELOPED_BY":   "MODEL/BENCHMARK -> ORGANIZATION that built it",
    "IS_A":           "specific thing -> more general category",
    "OUTPERFORMS":    "MODEL/METHOD -> MODEL/METHOD it is explicitly said to beat",
    "EVALUATES":      "BENCHMARK -> MODEL it tests (HELM EVALUATES GPT-3)",
    "COMPARED_WITH":  "A -> B when the text compares them without a clear winner",
    "EVALUATED_ON":   "MODEL -> BENCHMARK/DATASET/TASK it was tested on",
    "MEASURES":       "BENCHMARK/METRIC -> METRIC/CONCEPT it measures",
    "COVERS":         "BENCHMARK/DATASET -> DOMAIN/TASK it spans ",
    "TRAINED_ON":     "MODEL -> DATASET it was trained on",
    "USES":           "MODEL/METHOD -> METHOD it uses",
    "BUILDS_ON":      "newer thing -> older thing it extends",
    "PART_OF":        "component -> the whole it belongs to. NOT for two things listed together",
    "HAS_CAPABILITY": "MODEL -> TASK/DOMAIN it is said to be able to do",
    "HAS_LIMITATION": "MODEL/METHOD -> CONCEPT describing its own weakness ",
    "HAS_RISK":       "MODEL/METHOD -> CONCEPT it could be misused for or cause",
}

# simple sanity rules: (relation, which end) -> allowed entity types
TYPE_RULES = {
    ("DEVELOPED_BY", "target"):   {"ORGANIZATION"},
    ("TRAINED_ON", "target"):     {"DATASET"},
    ("EVALUATED_ON", "target"):   {"BENCHMARK", "DATASET", "TASK"},
    ("HAS_CAPABILITY", "source"): {"MODEL", "METHOD"},
    ("HAS_CAPABILITY", "target"): {"TASK", "DOMAIN"},
    ("OUTPERFORMS", "source"):    {"MODEL", "METHOD"},
    ("EVALUATES", "source"):      {"BENCHMARK"},
    ("EVALUATES", "target"):      {"MODEL"},
    ("COVERS", "source"):         {"BENCHMARK", "DATASET"},
    ("COVERS", "target"):         {"DOMAIN", "TASK"},
    ("HAS_LIMITATION", "source"): {"MODEL", "METHOD"},
    ("HAS_RISK", "source"):       {"MODEL", "METHOD"},
}
```

```python
class Entity(BaseModel):
    name: str = Field(description="Full name, singular, as written in the passage (no acronym if the full name is given)")
    entity_type: Literal["MODEL", "ORGANIZATION", "BENCHMARK", "DATASET", "METHOD",
                         "METRIC", "TASK", "DOMAIN", "CONCEPT"]
    aliases: list[str] = Field(default_factory=list, description="Other names used in the passage, e.g. acronyms like LLM, AGI")


class Relationship(BaseModel):
    source: str = Field(description="Entity name (must be one of the extracted entities)")
    relation: Literal["DEVELOPED_BY", "IS_A", "OUTPERFORMS", "EVALUATES", "COMPARED_WITH", "EVALUATED_ON",
                      "MEASURES", "COVERS", "TRAINED_ON", "USES", "BUILDS_ON", "PART_OF",
                      "HAS_CAPABILITY", "HAS_LIMITATION", "HAS_RISK"]
    target: str = Field(description="Entity name (must be one of the extracted entities)")
    evidence: str = Field(description="ONE sentence or clause copied exactly from the passage that states this relation")


class ChunkExtraction(BaseModel):
    entities: list[Entity]
    relationships: list[Relationship]
```

1. a passage with good relations,
2. a passage where the right answer is a **list of capabilities** (not `IS_A`),
3. "examples include X, Y" → `IS_A`, a benchmark that `EVALUATES` a model, a misuse risk, and two traps: citations and "X and Y" lists,
4. a passage with **no** relations at all.

## 4. Extract one chunk

`format=` passes the JSON schema to Ollama, so the answer always has the right shape. `temperature=0` makes it repeatable.

```python
def extract_chunk(chunk, timeout=300, max_tokens=None):
    user_prompt = f"""Paper: {chunk["title"]}
Section: {" > ".join(chunk["section_path"])}

Passage: {chunk["text"]}
Output:"""
    options = {"temperature": 0, "num_ctx": 8192}
    if max_tokens:
        options["num_predict"] = max_tokens        # cap the output, so a model stuck in a loop stops (section 16)

    response = requests.post(
        OLLAMA_URL,
        json={
            "model": MODEL_NAME,
            "stream": False,
            "format": ChunkExtraction.model_json_schema(),
            "options": options,
            "messages": [
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": user_prompt},
            ],
        },
        timeout=timeout,
    )
    response.raise_for_status()
    return ChunkExtraction.model_validate_json(response.json()["message"]["content"])
```

## 5. Normalize names

The same thing gets written many ways (`LLMs`, `large language models`, `Large language model`).
`canonical()` turns every spelling into **one** name:
1. tidy up the text: spaces, `GPT--4` → `GPT-4`
2. look it up in `ALIASES` (lowercase keys)
3. otherwise keep the name as written. Matching is case-insensitive, so "Artificial intelligence" = "artificial intelligence".

`ALIASES` starts with a few known names. The model's `aliases` field adds more for each chunk.

```python
ALIASES = {
    "llm": "large language model",
    "llms": "large language model",
    "large language models": "large language model",
    "language models": "language model",
    "lm": "language model",
    "agi": "artificial general intelligence",
    "ai": "artificial intelligence",
    "helm": "Holistic Evaluation of Language Models",
    "foundation models": "foundation model",
    "gpt4": "GPT-4",
    "gpt-4 model": "GPT-4",
    "rlhf": "reinforcement learning from human feedback",
    "cot": "chain-of-thought prompting",
}


def tidy(text):
    text = text.replace("“", '"').replace("”", '"').replace("’", "'").replace("‘", "'")
    text = text.replace("--", "-").replace("–", "-").replace("—", "-")
    return re.sub(r"\s+", " ", text).strip()


def canonical(name, extra_aliases=None):
    name = tidy(name)
    key = name.lower()
    if extra_aliases and key in extra_aliases:
        return extra_aliases[key]
    return ALIASES.get(key, name)


def key_of(name):
    """Case-insensitive key used to compare names."""
    return canonical(name).lower()
```

## 6. Validate every relationship

A relationship is **accepted** only if all checks pass. Otherwise it's **rejected** with a reason, so you can see *why*.

| Check | Catches |
|---|---|
| entity name appears in the passage (otherwise the entity is dropped) | invented entities |
| source and target are extracted entities | made-up endpoints like "early writings" |
| source ≠ target | self-loops |
| entity is not a citation ("et al.", "[..]") | authors and references as entities |
| quote found in the passage (after tidying quotes, dashes and **removing citation markers** like `[362]`) | invented quotes |
| at least one name is **in the quote**, and **both** are in the quote's sentence | quotes that don't talk about the pair, while still allowing "GPT-4 … passing a simulated bar exam" |
| type rules (`DEVELOPED_BY` → ORGANIZATION, `EVALUATES` → MODEL, ORGANIZATION only in `DEVELOPED_BY`…) | "GPT-4 DEVELOPED_BY mathematics", "Restaurant PART_OF Mei et al." |
| one relation per entity pair per chunk | "GPT-4 IS_A LLM" + "GPT-4 PART_OF LLM" duplicates |

```python
CITATION = re.compile(r"\[[^\]]{1,40}\]"                       # [362], [Ope23], [Devlin et al. 2019]
                      r"|\((?:[^()]*et al\.?[^()]*|[^()]*\b(?:19|20)\d\d[a-z]?)\)")   # (Konda et al., 2016), (2021)


def for_matching(text):
    """Text used for quote matching: tidy, no citation markers, lowercase."""
    text = CITATION.sub("", tidy(text))
    return re.sub(r"\s+", " ", text).strip().lower()


def is_citation(name):
    return "et al" in name.lower() or bool(re.search(r"\[|\]", name))


def sentence_around(passage, start, end):
    """The full sentence(s) that contain passage[start:end]."""
    left = max(passage.rfind(". ", 0, start), passage.rfind("? ", 0, start), passage.rfind("! ", 0, start))
    left = 0 if left == -1 else left + 2
    # search from the quote's last character, so a quote ending in "." stops at its own sentence
    ends = [i for i in (passage.find(". ", end - 1), passage.find("? ", end - 1), passage.find("! ", end - 1)) if i != -1]
    right = min(ends) + 1 if ends else len(passage)
    return passage[left:right]


def mentions(quote_key, entity_name, entity_aliases):
    """True if the entity's name, an alias, or a known ALIASES spelling appears in the quote
    as a whole word (an optional plural "s" is allowed, so "LLMs" matches "LLM")."""
    names = [entity_name] + entity_aliases
    names += [k for k, v in ALIASES.items() if v.lower() == entity_name.lower()]
    for n in names:
        pattern = r"(?<![a-z0-9])" + re.escape(tidy(n).lower()) + r"s?(?![a-z0-9])"
        if n and re.search(pattern, quote_key):
            return True
    return False


def validate(chunk, extraction):
    passage_key = for_matching(chunk["text"])

    # entity lookup: canonical key -> entity info (with aliases from this chunk)
    chunk_aliases = {}
    for e in extraction.entities:
        for a in e.aliases:
            chunk_aliases[tidy(a).lower()] = canonical(e.name)
    entities = {}
    for e in extraction.entities:
        if is_citation(e.name):                                # "Devlin et al." is not an entity
            continue
        if not mentions(tidy(chunk["text"]).lower(), e.name, e.aliases):   # entity must really be in the passage
            continue
        name = canonical(e.name, chunk_aliases)
        entities[name.lower()] = {"name": name, "type": e.entity_type,
                                  "aliases": [e.name] + e.aliases}

    seen_pairs = set()
    accepted, rejected = [], []
    for r in extraction.relationships:
        src = entities.get(canonical(r.source, chunk_aliases).lower())
        tgt = entities.get(canonical(r.target, chunk_aliases).lower())
        quote_key = for_matching(r.evidence)
        position = passage_key.find(quote_key) if quote_key else -1
        sentence = sentence_around(passage_key, position, position + len(quote_key)) if position >= 0 else ""

        if src is None:
            reason = "source not in entities"
        elif tgt is None:
            reason = "target not in entities"
        elif src["name"] == tgt["name"]:
            reason = "source == target"
        elif not quote_key:
            reason = "empty quote"
        elif position < 0:
            reason = "quote not in passage"
        elif not (mentions(quote_key, src["name"], src["aliases"]) or mentions(quote_key, tgt["name"], tgt["aliases"])):
            reason = "neither entity is in the quote"
        elif not mentions(sentence, src["name"], src["aliases"]):
            reason = "source not in the quote's sentence"
        elif not mentions(sentence, tgt["name"], tgt["aliases"]):
            reason = "target not in the quote's sentence"
        elif "ORGANIZATION" in (src["type"], tgt["type"]) and r.relation != "DEVELOPED_BY":
            reason = f"ORGANIZATION only allowed in DEVELOPED_BY, not {r.relation}"
        elif src["type"] not in TYPE_RULES.get((r.relation, "source"), {src["type"]}):
            reason = f"source type {src['type']} not allowed for {r.relation}"
        elif tgt["type"] not in TYPE_RULES.get((r.relation, "target"), {tgt["type"]}):
            reason = f"target type {tgt['type']} not allowed for {r.relation}"
        elif (src["name"], tgt["name"]) in seen_pairs:
            reason = "duplicate: this pair already has a relation in this chunk"
        else:
            reason = None
            seen_pairs.add((src["name"], tgt["name"]))

        row = {
            "source": src["name"] if src else r.source,
            "relation": r.relation,
            "target": tgt["name"] if tgt else r.target,
            "evidence": r.evidence,
            "chunk_id": chunk["chunk_id"],
            "document": chunk["document"],
            "section": " > ".join(chunk["section_path"]),
            "page": chunk["page_start"],
        }
        if reason:
            rejected.append({**row, "reason": reason})
        else:
            accepted.append(row)

    entity_rows = [{"name": v["name"], "type": v["type"], "chunk_id": chunk["chunk_id"],
                    "document": chunk["document"]} for v in entities.values()]
    return entity_rows, accepted, rejected
```

## 8. Run extraction with checkpointing

Each chunk's result is **appended to a `.jsonl` file as soon as it's done**. If the run stops (crash, closed laptop, Ollama error), just run the cell again: it **skips chunks already in the file** and **retries chunks that errored**.
This same function will process all ~2,900 chunks later; only the list and the file name change.

```python
def load_done(path):
    if not path.exists():
        return {}
    with path.open(encoding="utf-8") as f:
        rows = [json.loads(line) for line in f if line.strip()]
    return {row["chunk_id"]: row for row in rows}


def run_extraction(chunk_list, output_path, **llm_options):
    """llm_options are passed to extract_chunk(), e.g. timeout=900, max_tokens=4096."""
    done = load_done(output_path)
    # skip chunks that are done; retry the ones that errored last time
    todo = [c for c in chunk_list if c["chunk_id"] not in done or done[c["chunk_id"]]["error"]]
    print(f"{len(chunk_list) - len(todo)} already done, {len(todo)} to go")

    for i, chunk in enumerate(todo, start=1):
        start = time.time()
        try:
            extraction = extract_chunk(chunk, **llm_options)
            entity_rows, accepted, rejected = validate(chunk, extraction)
            row = {"chunk_id": chunk["chunk_id"], "entities": entity_rows,
                   "accepted": accepted, "rejected": rejected, "error": None}
        except Exception as exc:
            row = {"chunk_id": chunk["chunk_id"], "entities": [],
                   "accepted": [], "rejected": [], "error": str(exc)}

        with output_path.open("a", encoding="utf-8") as f:     # checkpoint: save right away
            f.write(json.dumps(row, ensure_ascii=False) + "\n")

        print(f"[{i}/{len(todo)}] {chunk['chunk_id']:25} "
              f"entities={len(row['entities']):2} accepted={len(row['accepted']):2} "
              f"rejected={len(row['rejected']):2} {time.time() - start:5.1f}s"
              + (f"  ERROR: {row['error'][:80]}" if row["error"] else ""))

    return load_done(output_path)
```

## 14. Normalize entity identities across papers

Each chunk was normalized on its own. Across 2,900 chunks the same thing still shows up in different spellings:
`GPT-4` / `GPT 4` / `gpt-4`, `large language model` / `large language models`, `chain-of-thought` / `chain of thought`.

We give every name an **identity key** and merge names with the same key:
1. `canonical()` + the `ALIASES` map (from section 5)
2. lowercase, and remove spaces, hyphens and dots, so `GPT-4` = `gpt 4` = `GPT4`
3. a plural is merged **only if the singular also exists** (`foundation models` → `foundation model`, but `bias` stays `bias`)
4. `MANUAL_MERGES` for anything the rules miss. Look at the check cells and add pairs there.

Each merged entity gets:
- **name**: its most common spelling
- **type**: its most common type
- **aliases**: all other spellings

```python
# add pairs you spot in the checks below:  "spelling to replace": "name to keep"
MANUAL_MERGES = {
    "gpt-4 (launch)": "GPT-4",
    "gpt-4 (early)": "GPT-4",
    "openai gpt-4": "GPT-4",
}


def squash(name):
    """Identity key: canonical, lowercase, no spaces/hyphens/dots/underscores."""
    name = MANUAL_MERGES.get(tidy(name).lower(), name)
    return re.sub(r"[\s\-_.]", "", canonical(name).lower())


# 1) key for every spelling
spellings = Counter(entities_df["name"].tolist() + facts_df["source"].tolist() + facts_df["target"].tolist())
key_of_name = {name: squash(name) for name in spellings}

# 2) plural -> singular, only when the singular exists
all_keys = set(key_of_name.values())
for name, key in key_of_name.items():
    if key.endswith("s") and not key.endswith("ss") and key[:-1] in all_keys:
        key_of_name[name] = key[:-1]

# 3) one display name, type and alias list per key
groups = {}
for name, count in spellings.items():
    groups.setdefault(key_of_name[name], Counter())[name] += count

type_votes = {}
for row in entities_df.itertuples():
    type_votes.setdefault(key_of_name[row.name], Counter())[row.type] += 1

entity_rows = []
for key, names in groups.items():
    display_name = canonical(names.most_common(1)[0][0])         # e.g. "HELM" -> full name from ALIASES
    for preferred in names:                                   # prefer the ALIASES / MANUAL_MERGES target spelling
        if preferred in set(ALIASES.values()) | set(MANUAL_MERGES.values()):
            display_name = preferred
    votes = type_votes.get(key, Counter({"CONCEPT": 1}))
    entity_rows.append({
        "key": key,
        "name": display_name,
        "type": votes.most_common(1)[0][0],
        "type_votes": dict(votes),
        "aliases": sorted(n for n in names if n != display_name),
        "mentions": sum(names.values()),
    })

entities_final = pd.DataFrame(entity_rows).set_index("key")
name_of_key = entities_final["name"].to_dict()
final_name = {name: name_of_key[key] for name, key in key_of_name.items()}

print(f"{len(spellings)} spellings -> {len(entities_final)} entities")
```

```python
# check: which spellings were merged? (read these - wrong merges are easy to spot here)
merged = entities_final[entities_final["aliases"].map(len) > 0].sort_values("mentions", ascending=False)
print(f"{len(merged)} entities have more than one spelling")
display(merged[["name", "type", "aliases", "mentions"]].head(30))
```

## 15. Final graph facts

Rename every fact's source and target to the merged entity names, then:
- **drop self-loops** that merging can create (`GPT 4 → GPT-4`)
- **merge duplicate facts**: the same `(source, relation, target)` found in many chunks becomes one fact, with a `support` count and all its quotes and chunk IDs. Facts found in several chunks or papers are the most trustworthy.
- keep only entities that appear in **at least one fact**. Loose entities ("image", "prompt") never enter the graph.

```python
facts_df["source_final"] = facts_df["source"].map(final_name)
facts_df["target_final"] = facts_df["target"].map(final_name)
facts_df = facts_df[facts_df["source_final"] != facts_df["target_final"]]          # no self-loops

graph_facts = (
    facts_df.groupby(["source_final", "relation", "target_final"])
    .agg(support=("chunk_id", "nunique"),
         documents=("document", lambda d: sorted(set(d))),
         chunk_ids=("chunk_id", lambda c: sorted(set(c))),
         evidence=("evidence", lambda e: list(dict.fromkeys(e))[:5]),     # up to 5 different quotes
         pages=("page", lambda p: sorted(set(p))))
    .reset_index()
    .rename(columns={"source_final": "source", "target_final": "target"})
    .sort_values("support", ascending=False)
)

used = set(graph_facts["source"]) | set(graph_facts["target"])
graph_entities = entities_final[entities_final["name"].isin(used)].reset_index()

print(f"facts: {len(facts_df)} raw -> {len(graph_facts)} unique")
print(f"entities: {len(entities_final)} total -> {len(graph_entities)} used in facts")
```

2. **More plural forms.** Besides `-s`, also `-es` (`biases` → `bias`) and `-ies` (`emergent abilities` → `emergent ability`). Still only when the singular exists.
3. **More aliases** (`EXTRA_ALIASES`), found by searching the v1 names for acronyms and spelling variants (`SFT`, `CoT reasoning`, `MLLMs`…).

```python
# found in the v1 names: "spelling" -> "name to keep" (lowercase keys, like ALIASES)
EXTRA_ALIASES = {
    "reinforcement learning with human feedback": "reinforcement learning from human feedback",
    "rlhf fine-tuning": "reinforcement learning from human feedback",
    "chain-of-thought": "chain-of-thought prompting",
    "chain of thought": "chain-of-thought prompting",
    "chain-of-thoughts": "chain-of-thought prompting",
    "chain-of-thought reasoning": "chain-of-thought prompting",
    "chain-of-thought (cot) prompting": "chain-of-thought prompting",
    "cot prompting": "chain-of-thought prompting",
    "cot reasoning": "chain-of-thought prompting",
    "icl": "in-context learning",
    "sft": "supervised fine-tuning",
    "supervised fine-tuning (sft)": "supervised fine-tuning",
    "peft": "parameter-efficient fine-tuning",
    "rag": "retrieval-augmented generation",
    "mllm": "multimodal large language model",
    "mllms": "multimodal large language model",
    "pii": "personally identifiable information",
    "personally identifying information": "personally identifiable information",
    "rm": "reward model",
    "nlp": "natural language processing",
}
ALIASES.update(EXTRA_ALIASES)        # canonical() and mentions() use them from now on

MANUAL_MERGES.update({               # same idea as "gpt-4 (early)" in section 14
    "gpt-4-early": "GPT-4",
    "gpt-4-launch": "GPT-4",
    "gpt-4 (no vision)": "GPT-4",
})

SIZE_SUFFIX = re.compile(r"\s*\(\s*~?\d+(?:\.\d+)?\s*[BMT]\s*\)$"     # "LLaMA (7B)", "ada (350M)"
                         r"|[\s-]\d+(?:\.\d+)?[BMT]$", re.IGNORECASE)   # "LLaMA-7B", "LLaMA 13B"
VERSION_DATE = re.compile(r"\s+v20\d{6}$")                              # "Cohere xlarge v20220609"


def strip_variant(name):
    """Drop a model size or version date at the end: 'LLaMA-2-7B' -> 'LLaMA-2'."""
    name = SIZE_SUFFIX.sub("", tidy(name))
    return VERSION_DATE.sub("", name).strip()


def squash_v2(name):
    """Identity key: like squash(), but sizes and version dates don't count."""
    name = strip_variant(name)
    name = MANUAL_MERGES.get(name.lower(), name)
    return re.sub(r"[\s\-_.]", "", canonical(name).lower())


def singular_key(key, all_keys):
    """'emergentabilities' -> 'emergentability', but only if that key exists."""
    candidates = []
    if key.endswith("ies"):
        candidates.append(key[:-3] + "y")          # abilities -> ability
    if key.endswith("es"):
        candidates.append(key[:-2])                # biases -> bias
    if key.endswith("s") and not key.endswith("ss"):
        candidates.append(key[:-1])                # models -> model
    for c in candidates:
        if len(c) >= 3 and c in all_keys:
            return c
    return key
```

```python
# check: these spellings should land on the same key as their partner
tests = ["LLaMA", "LLaMA (7B)", "LLaMA-7B", "LLaMA 2 (7B)", "LLaMA-2-7B",
         "Cohere xlarge v20220609 (52.4B)", "Cohere xlarge", "GPT-4-early", "GPT-4",
         "SFT", "supervised fine-tuning", "CoT reasoning", "chain-of-thought prompting", "MLLMs"]
for t in tests:
    print(f"{t:35} -> {squash_v2(t)}")
demo_keys = {"emergentability", "bias", "model", "glass"}
print()
for k in ["emergentabilities", "biases", "models", "glasses", "class", "series"]:
    print(f"{k:20} -> {singular_key(k, demo_keys)}")
```

## 18. Junk entities

Names that should never be graph nodes:
- the **type written as a name** (`CONCEPT`, `METHOD`): the model put the type in the name field
- **references to the paper itself**: `Table 5`, `Figure 7.2`, `Appendix A`
- **citations** (`Henighan et al.`, `[Ope23]`), using `is_citation()` from section 6
- **no letters** (`2020`, `80%`), a single character, only a size (`(1.3B)`), or a **URL**
- **generic words** in `GENERIC_NAMES` (`scenario`, `prompt`, `models`…). The prompt told the model to skip these, but it didn't always.

A fact with a junk name at either end is dropped, v1 facts included. Add words to `GENERIC_NAMES` if you spot more in the check.

```python
GENERIC_NAMES = {
    "model", "models", "system", "systems", "paper", "papers", "work", "study", "approach", "approaches",
    "method", "methods", "result", "results", "task", "tasks", "data", "dataset", "datasets",
    "example", "examples", "experiment", "experiments", "scenario", "scenarios", "table", "figure",
    "text", "input", "inputs", "output", "outputs", "prompt", "prompts", "performance",
    "limitation", "limitations", "use case", "use cases", "user", "users", "people", "researchers",
    "authors", "downstream tasks", "training data", "it", "this", "that", "they", "we", "an", "a", "the",
}
REFERENCE = re.compile(r"^(appendix|section|sec\.|table|figure|fig\.|chapter|equation|eq\.|footnote)"
                       r"\s*([a-z]|\d+(\.\d+)*)$", re.IGNORECASE)


def is_junk(name):
    name = strip_variant(name)                  # "(1.3B)" -> "" counts as junk too
    return (len(name) < 2
            or not re.search(r"[A-Za-z]", name)
            or is_citation(name)
            or bool(re.search(r"https?://|www\.", name))
            or bool(REFERENCE.match(name))
            or name in ENTITY_TYPES             # "CONCEPT", "METHOD"
            or name.lower() in GENERIC_NAMES)
```

## 19. Re-validate the rejected relations

v1 saved every rejected relation with its reason. Here's what random samples of each reason showed:

| v1 reason | count | What the samples showed | v2 |
|---|---|---|---|
| source / target not in entities | ~5,200 | Some are real names the model forgot to list (`BMTrain DEVELOPED_BY OpenBMB`). Others are guesses, like `GPT-4` in a chunk that never names GPT-4 | **Second check**: the name must appear in the passage, then the relation must pass every other v1 check |
| quote not in passage | ~1,200 | Paraphrased or stitched-together quotes, many from tables | Keep rejecting: this is what stops made-up facts |
| neither name, or one name, not in the quote's sentence | ~3,200 | Mostly guesses from context (`XSUM HAS_CAPABILITY sunbathing`) | Keep rejecting |
| type rules | ~1,500 | Mostly wrong relations (`GPT-4 HAS_CAPABILITY governance`, table headers as metrics) | Keep rejecting |

So v2 only loosens the bookkeeping check. The other checks were rejecting mostly bad relations, so they stay.

For the second check, every name gets:
- its **identity key** from section 17, so all spellings of it count when looking for it in the passage, quote and sentence
- its **corpus-wide type** (majority vote over all chunks), because a name the model forgot has no type of its own in that chunk.
  If the name was **never** typed anywhere and the relation has a type rule for that end (`DEVELOPED_BY` → ORGANIZATION…), the relation is rejected: we can't check it.

```python
# 1) identity key for every spelling (junk left out), with the plural rule
squashed = {name: squash_v2(name) for name in name_counts if not is_junk(name)}
all_keys = set(squashed.values())
key_v2 = {name: singular_key(key, all_keys) for name, key in squashed.items()}

variants = {}                          # key -> every spelling seen for it
for name, key in key_v2.items():
    variants.setdefault(key, set()).add(name)

# 2) corpus-wide type: majority vote over every chunk the entity was extracted from
type_votes_v2 = {}
for r in ok_results.values():
    for e in r["entities"]:
        if e["name"] in key_v2:
            type_votes_v2.setdefault(key_v2[e["name"]], Counter())[e["type"]] += 1
corpus_type = {key: votes.most_common(1)[0][0] for key, votes in type_votes_v2.items()}

print(f"{len(name_counts)} spellings -> {len(variants)} identity keys ({len(junk_counts)} junk spellings left out)")
```

```python
SECOND_CHECK = {"source not in entities", "target not in entities"}


def check_v2(row):
    """v1's checks again, but a name only has to appear in the passage (not in the entity list).
    Returns the reason for rejecting, or None if the relation is accepted."""
    chunk = chunk_by_id[row["chunk_id"]]
    src, tgt = row["source"], row["target"]
    if is_junk(src) or is_junk(tgt):
        return "junk entity"
    src_key, tgt_key = key_v2[src], key_v2[tgt]
    if src_key == tgt_key:
        return "source == target"

    src_names, tgt_names = sorted(variants[src_key]), sorted(variants[tgt_key])
    passage = tidy(chunk["text"]).lower()
    if not mentions(passage, src, src_names):
        return "source not named in the passage"
    if not mentions(passage, tgt, tgt_names):
        return "target not named in the passage"

    passage_key = for_matching(chunk["text"])
    quote_key = for_matching(row["evidence"])
    position = passage_key.find(quote_key) if quote_key else -1
    if position < 0:
        return "quote not in passage"
    if not (mentions(quote_key, src, src_names) or mentions(quote_key, tgt, tgt_names)):
        return "neither entity is in the quote"
    sentence = sentence_around(passage_key, position, position + len(quote_key))
    if not mentions(sentence, src, src_names):
        return "source not in the quote's sentence"
    if not mentions(sentence, tgt, tgt_names):
        return "target not in the quote's sentence"

    relation = row["relation"]
    src_type, tgt_type = corpus_type.get(src_key), corpus_type.get(tgt_key)     # None = never typed
    if "ORGANIZATION" in (src_type, tgt_type) and relation != "DEVELOPED_BY":
        return f"ORGANIZATION only allowed in DEVELOPED_BY, not {relation}"
    for end, end_type in (("source", src_type), ("target", tgt_type)):
        allowed = TYPE_RULES.get((relation, end))
        if allowed and end_type is None:
            return f"{end} never typed, can't check the {relation} type rule"
        if allowed and end_type not in allowed:
            return f"{end} type {end_type} not allowed for {relation}"
    return None
```

```python
raw_v2, rejected_v2 = [], []
for cid, r in ok_results.items():
    seen_pairs = set()                     # one relation per entity pair per chunk, like v1
    for row in r["accepted"]:              # v1 facts stay, unless an end is junk
        if is_junk(row["source"]) or is_junk(row["target"]):
            rejected_v2.append({**row, "reason": "junk entity (was accepted in v1)"})
            continue
        seen_pairs.add((key_v2[row["source"]], key_v2[row["target"]]))
        raw_v2.append({**row, "origin": "v1"})

    for row in r["rejected"]:
        if row["reason"] not in SECOND_CHECK:
            continue
        reason = check_v2(row)
        pair = None if reason else (key_v2[row["source"]], key_v2[row["target"]])
        if pair in seen_pairs:
            reason = "duplicate: this pair already has a relation in this chunk"
        if reason:
            rejected_v2.append({**row, "reason": reason})
        else:
            seen_pairs.add(pair)
            raw_v2.append({**row, "origin": "new"})

raw_v2_df = pd.DataFrame(raw_v2)
print("v1 accepted:", sum(len(r["accepted"]) for r in ok_results.values()))
print(raw_v2_df["origin"].value_counts().to_string())
print("total v2 relations:", len(raw_v2_df))
```

```python
# check: what happened to the relations that got a second check?
def reason_group(reason):
    if "never typed" in reason:
        return "a name was never typed, so its type rule can't be checked"
    if reason.startswith(("source type", "target type", "ORGANIZATION")):
        return "type rules"
    return reason


second = Counter(reason_group(x["reason"]) for x in rejected_v2 if x["reason"] != "junk entity (was accepted in v1)")
second["ACCEPTED"] = int((raw_v2_df["origin"] == "new").sum())
display(pd.Series(second).sort_values(ascending=False).rename("relations").to_frame())
print("v1 facts dropped as junk:", sum(1 for x in rejected_v2 if x["reason"] == "junk entity (was accepted in v1)"))
```

## 20. Build the v2 graph

Same steps as sections 14–15, with the v2 keys:
- every key gets one **name** (the most common singular spelling, without size), one **type** (majority vote) and its other spellings as **aliases**
- facts are renamed to those names, self-loops dropped, duplicates merged with a `support` count
- only entities used in at least one fact are kept
- every fact keeps an `origin`: `v1` if v1 already had it, `new` if only the second check found it.
  If your grading in section 22 shows the `new` facts are much worse, the query notebook can leave them out.

```python
# one name, type and alias list per key
spellings_by_key = {}
for name, key in key_v2.items():
    spellings_by_key.setdefault(key, Counter())[name] += name_counts[name]

preferred = set(ALIASES.values()) | set(MANUAL_MERGES.values())
entity_rows_v2 = []
for key, names in spellings_by_key.items():
    singular = Counter({n: c for n, c in names.items() if squash_v2(n) == key}) or names   # not a plural spelling
    display_name = canonical(strip_variant(singular.most_common(1)[0][0]))
    for name in names:                                 # prefer the ALIASES / MANUAL_MERGES spelling
        if name in preferred:
            display_name = name
    votes = type_votes_v2.get(key, Counter({"CONCEPT": 1}))
    entity_rows_v2.append({
        "key": key,
        "name": display_name,
        "type": votes.most_common(1)[0][0],
        "type_votes": dict(votes),
        "aliases": sorted(n for n in names if n != display_name),
        "mentions": sum(names.values()),
    })
entities_v2 = pd.DataFrame(entity_rows_v2).set_index("key")
name_of_key_v2 = entities_v2["name"].to_dict()

# facts: rename, drop self-loops, merge duplicates
facts_v2 = raw_v2_df.copy()
facts_v2["source"] = facts_v2["source"].map(lambda n: name_of_key_v2[key_v2[n]])
facts_v2["target"] = facts_v2["target"].map(lambda n: name_of_key_v2[key_v2[n]])
facts_v2 = facts_v2[facts_v2["source"] != facts_v2["target"]]

graph_facts_v2 = (
    facts_v2.groupby(["source", "relation", "target"])
    .agg(support=("chunk_id", "nunique"),
         documents=("document", lambda d: sorted(set(d))),
         chunk_ids=("chunk_id", lambda c: sorted(set(c))),
         evidence=("evidence", lambda e: list(dict.fromkeys(e))[:5]),
         pages=("page", lambda p: sorted(set(p))),
         origin=("origin", lambda o: "v1" if "v1" in set(o) else "new"))
    .reset_index()
    .sort_values("support", ascending=False)
)
used_v2 = set(graph_facts_v2["source"]) | set(graph_facts_v2["target"])
graph_entities_v2 = entities_v2[entities_v2["name"].isin(used_v2)].reset_index()

print(f"facts: {len(facts_v2)} raw -> {len(graph_facts_v2)} unique "
      f"({(graph_facts_v2['origin'] == 'new').sum()} only from the second check)")
print(f"entities: {len(entities_v2)} total -> {len(graph_entities_v2)} used in facts")
```

```python
# check: names that now merge sizes, plurals or new aliases (read these - wrong merges are easy to spot here)
# (a spelling whose v1 key was different, and not just the v1 "-s" plural rule)
def merged_only_in_v2(row):
    v1_keys = {squash(row["name"]), squash(row["name"]) + "s"}
    return [a for a in row["aliases"] if squash(a) not in v1_keys and squash(a)[:-1] not in v1_keys]


graph_entities_v2["new_aliases"] = graph_entities_v2.apply(merged_only_in_v2, axis=1)
new_merges = graph_entities_v2[graph_entities_v2["new_aliases"].map(len) > 0]
print(f"{len(new_merges)} entities merged spellings that v1 kept apart")
display(new_merges.sort_values("mentions", ascending=False)[["name", "type", "new_aliases", "mentions"]].head(25))
graph_entities_v2 = graph_entities_v2.drop(columns="new_aliases")
```

```python
# save the v2 graph next to v1
assert graph_entities_v2["name"].is_unique, "two entities got the same name - check MANUAL_MERGES / EXTRA_ALIASES"
FACTS_V2_PATH = ARTIFACTS_DIR / "graph_facts_v2.jsonl"
ENTITIES_V2_PATH = ARTIFACTS_DIR / "graph_entities_v2.csv"

with FACTS_V2_PATH.open("w", encoding="utf-8") as f:
    for row in graph_facts_v2.to_dict("records"):
        f.write(json.dumps(row, ensure_ascii=False) + "\n")
graph_entities_v2.to_csv(ENTITIES_V2_PATH, index=False, encoding="utf-8-sig")

# check: reload and verify every fact points to a known entity
reloaded = [json.loads(line) for line in FACTS_V2_PATH.open(encoding="utf-8")]
known = set(pd.read_csv(ENTITIES_V2_PATH)["name"])
bad = [f for f in reloaded if f["source"] not in known or f["target"] not in known]
print("saved:", FACTS_V2_PATH.name, f"({len(reloaded)} facts) |", ENTITIES_V2_PATH.name, f"({len(known)} entities)")
assert not bad, "some facts point to entities that are missing from graph_entities_v2.csv"
```

## 21. Compare v1 and v2

The same numbers for both graphs, read from the saved files.
The last table counts facts that touch each topic of the test questions. A topic with 0–3 facts means GraphRAG will have to answer it from the chunk text alone.

```python
def load_graph(facts_path):
    facts = [json.loads(line) for line in facts_path.open(encoding="utf-8")]
    degree, papers_of = Counter(), {}
    for f in facts:
        for name in (f["source"], f["target"]):
            degree[name] += 1
            papers_of.setdefault(name, set()).update(f["documents"])
    return facts, degree, papers_of


def graph_summary(facts_path):
    facts, degree, papers_of = load_graph(facts_path)
    return {
        "facts": len(facts),
        "facts only from the second check": sum(f.get("origin") == "new" for f in facts),
        "entities": len(degree),
        "chunks with a fact": len({c for f in facts for c in f["chunk_ids"]}),
        "facts from 2+ chunks": sum(f["support"] >= 2 for f in facts),
        "facts from 2+ papers": sum(len(f["documents"]) >= 2 for f in facts),
        "entities in 2+ papers": sum(len(p) >= 2 for p in papers_of.values()),
        "entities with 1 connection": f"{sum(d == 1 for d in degree.values()) / len(degree):.0%}",
    }


v1_path, v2_path = ARTIFACTS_DIR / "graph_facts.jsonl", FACTS_V2_PATH
display(pd.DataFrame({"v1": graph_summary(v1_path), "v2": graph_summary(v2_path)}))

per_paper = pd.DataFrame({
    version: Counter(d for f in load_graph(path)[0] for d in f["documents"])
    for version, path in [("v1", v1_path), ("v2", v2_path)]})
display(per_paper.T)
```

```python
TOPICS = ["calibration", "scaling", "homogenization", "theory of mind", "bar exam",
          "hallucination", "emergen", "rlhf|human feedback", "chain-of-thought"]


def topic_facts(facts_path, pattern):
    facts = load_graph(facts_path)[0]
    return sum(bool(re.search(pattern, f["source"] + " | " + f["target"], re.IGNORECASE)) for f in facts)


display(pd.DataFrame({"v1": {t: topic_facts(v1_path, t) for t in TOPICS},
                      "v2": {t: topic_facts(v2_path, t) for t in TOPICS}}))
```

The v1 precision (71%) came from 17 relations, all from the 15 test chunks. This samples 50 from the full run:
**25 that v1 already accepted** (`origin = v1`) and **25 that only v2 accepts** (`origin = new`), so you can see whether the second check lets in worse relations.

Same as section 10: open `artifacts/review_graph_v2.csv` in Excel and fill `verdict` with `correct`, `partial` or `wrong`. Then run the cell after it.
Tip: grade a row before you look at its `origin`, so it doesn't bias you.

```python
REVIEW_V2_PATH = ARTIFACTS_DIR / "review_graph_v2.csv"

sample_v2 = pd.concat([
    raw_v2_df[raw_v2_df["origin"] == "v1"].sample(25, random_state=42),
    raw_v2_df[raw_v2_df["origin"] == "new"].sample(25, random_state=42),
]).sample(frac=1, random_state=42)                        # shuffle, so the two groups are mixed
review_v2 = sample_v2[["chunk_id", "source", "relation", "target", "evidence", "origin"]].copy()
review_v2["verdict"] = ""
if not REVIEW_V2_PATH.exists():                          # never overwrite your grading
    review_v2.to_csv(REVIEW_V2_PATH, index=False, encoding="utf-8-sig")
    print("written:", REVIEW_V2_PATH)
else:
    print("already exists (not overwritten):", REVIEW_V2_PATH)
```

```python
# check: precision from your grading, for v1 relations, new relations, and the whole v2 graph
graded = pd.read_csv(REVIEW_V2_PATH)
graded["verdict"] = graded["verdict"].fillna("").str.strip().str.lower()
done = graded[graded["verdict"] != ""]
print(f"graded {len(done)} of {len(graded)}")

if len(done):
    by_origin = done.groupby("origin")["verdict"].agg(
        graded="size",
        strict=lambda v: (v == "correct").mean(),
        lenient=lambda v: v.isin(["correct", "partial"]).mean())
    display(by_origin.assign(strict=by_origin["strict"].map("{:.0%}".format),
                             lenient=by_origin["lenient"].map("{:.0%}".format)))

    # whole graph = each group's precision weighted by its share of all v2 relations
    share = raw_v2_df["origin"].value_counts(normalize=True)
    if set(by_origin.index) == set(share.index):
        overall_strict = sum(by_origin.loc[o, "strict"] * share[o] for o in share.index)
        overall_lenient = sum(by_origin.loc[o, "lenient"] * share[o] for o in share.index)
        print(f"v2 graph estimate: strict {overall_strict:.0%} | lenient {overall_lenient:.0%}")
    display(done[done["verdict"] == "wrong"][["origin", "source", "relation", "target", "evidence"]])
```
