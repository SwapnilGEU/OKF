## 1. Setup

The only real decision here is **the judge**: the model that reads each answer and context and says yes/no to DeepEval's questions.

| `JUDGE_PROVIDER` | `JUDGE_MODEL` | notes |
|---|---|---|
| `"mercury"` | `"mercury-2.5"` | **current judge.** Inception's diffusion LLM through its API; needs `INCEPTION_API_KEY` in `.env`. A different model family from the answering model, very fast, and it can return JSON in the exact shape DeepEval asks for. |
| `"ollama"` | `"qwen3:4b-instruct-2507-q4_K_M"` | free, already installed. But it's the same model that wrote the answers, so it's grading its own work, and a 4B model misses subtle mistakes. |
| `"ollama"` | `"qwen3:8b"` | `ollama pull qwen3:8b` first. Bigger and different from the answering model. With 6 GB VRAM part of it runs on the CPU, so it's slower. |
| `"openai"` | `"gpt-4.1-mini"` (or similar) | needs `OPENAI_API_KEY` in `.env`. DeepEval's built-in judge. |

**Mercury's budget.** One question costs about **28 judge calls** (4 methods × ~7) of roughly 5,000 tokens each, so the whole set is about **5–6 M tokens**. The free 100 M tokens cover that many times over, re-runs included. Calls are billed per token, so the notebook adds up the tokens it uses and prints them after the scoring run.

**`MERCURY_REASONING`**: Mercury 2.5 thinks before answering, and the thinking is billed as output tokens. `"low"` is enough for yes/no verdicts; `"medium"` is slower and uses more tokens but may catch subtler errors.

**Where the key goes:** in `.env` as `INCEPTION_API_KEY=...` (`.env` is in `.gitignore`). Don't paste it into a notebook: the notebooks go to GitHub.

Whatever you pick, **use the same judge for all four methods**. Scores from different judges aren't comparable, which is why `scores.jsonl` keeps the judge's name on every line (the old qwen and Gemini scores stay in there under their own names).

`QUESTION_LIMIT = 3` is a good first run, to check the judge before starting all of them.

````python
import asyncio
import hashlib
import json
import os
import re
import time
from pathlib import Path

os.environ.setdefault("DEEPEVAL_TELEMETRY_OPT_OUT", "YES")

import pandas as pd
import requests
from dotenv import load_dotenv
from deepeval.metrics import (AnswerRelevancyMetric, ContextualPrecisionMetric,
                              ContextualRecallMetric, FaithfulnessMetric)
from deepeval.models import DeepEvalBaseLLM
from deepeval.test_case import LLMTestCase

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
load_dotenv(PROJECT_ROOT / ".env")

EVAL_PATH = PROJECT_ROOT / "data" / "eval" / "eval_questions.jsonl"
RESULTS_DIR = PROJECT_ROOT / "artifacts" / "results"
SCORES_PATH = RESULTS_DIR / "scores.jsonl"

METHODS = {                      # results file name -> label in the tables
    "okf": "OKF",
    "okf_hybrid": "OKF hybrid",
    "graph": "Graph",
    "graph_hybrid": "Graph hybrid",
}

# the judge: one for every method
# JUDGE_PROVIDER = "ollama"                       # the local 4B judge
# JUDGE_MODEL = "qwen3:4b-instruct-2507-q4_K_M"   # or "qwen3:8b"
JUDGE_PROVIDER = "mercury"                        # "mercury", "ollama" or "openai"
JUDGE_MODEL = "mercury-2.5"
INCEPTION_API_KEY = os.getenv("INCEPTION_API_KEY", "")   # in .env (git-ignored); this notebook goes to GitHub
MERCURY_REASONING = "low"                         # "low" / "medium" / "high": more thinking = more output tokens

OLLAMA_URL = "http://localhost:11434"
JUDGE_NUM_CTX = 12288        # contexts are up to ~18,000 characters (~4,500 tokens) plus DeepEval's instructions
JUDGE_MAX_TOKENS = 4096      # cap per judge reply, so a model stuck repeating itself can't run forever

INCLUDE_REASON = False       # True = the judge also writes a sentence explaining each score (one more call per metric)
THRESHOLD = 0.5              # only used for DeepEval's pass/fail flag; the tables use the raw scores
QUESTION_LIMIT = None        # e.g. 3 for a quick trial run

RESULTS_DIR.mkdir(parents=True, exist_ok=True)
print("project:", PROJECT_ROOT)
print("judge:", JUDGE_PROVIDER, JUDGE_MODEL)
````

## 2. Load the questions and the four results files

A results line counts only if its question text still matches the questions file. If you edit a question (or add new ones), re-run the save-results cell in each query notebook, otherwise that question shows up as **stale** or **missing** below.

````python
questions = [json.loads(line) for line in EVAL_PATH.open(encoding="utf-8") if line.strip()]
by_id = {q["id"]: q for q in questions}
assert len(by_id) == len(questions), "question ids must be unique"

results = {}
for method in METHODS:
    path = RESULTS_DIR / f"{method}.jsonl"
    if path.exists():
        results[method] = {r["id"]: r for r in (json.loads(line) for line in path.open(encoding="utf-8") if line.strip())}
    else:
        results[method] = {}
        print(f"missing {path.name}: run the save-results cell in that notebook")


def result_for(method, q):
    r = results[method].get(q["id"])
    return r if r is not None and r["question"] == q["question"] else None


def status(method, q):
    if q["id"] not in results[method]:
        return "missing"
    return "ok" if result_for(method, q) else "stale"


coverage = pd.DataFrame({label: [status(m, q) for q in questions] for m, label in METHODS.items()},
                        index=[q["id"] for q in questions])
print(f"{len(questions)} questions: {sum(q['answerable'] for q in questions)} answerable, "
      f"{sum(not q['answerable'] for q in questions)} unanswerable")
print(pd.Series([q["type"] for q in questions]).value_counts().to_string(), "\n")
print(coverage.apply(pd.Series.value_counts).fillna(0).astype(int).to_string())
problems = coverage[(coverage != "ok").any(axis=1)]
if len(problems):
    print("\nnot ok:\n" + problems.to_string())
````

## 3. The four metrics

Two of them grade **retrieval** (did the method find the right material?) and two grade **generation** (did the model use it well?). That split is what tells you *why* a method loses: it never found the facts, or it found them and wrote a poor answer.

| metric | grades | uses | what the judge is asked |
|---|---|---|---|
| **Contextual precision** | retrieval | question, reference answer, context blocks | for each block, in order: is it useful for the reference answer? The score is high when the useful blocks come **first**. |
| **Contextual recall** | retrieval | reference answer, context blocks | for each sentence of the reference answer: can it be found in the context? |
| **Faithfulness** | generation | answer, context blocks | pull the claims out of the answer; is each one supported by the context? |
| **Answer relevancy** | generation | question, answer | split the answer into statements; does each one address the question? |

All four are 0–1, higher is better. Each one is a few judge calls (extract, then give verdicts), so one question × one method ≈ 6–10 calls.

Two things these four **don't** do:
- None compares the answer with the reference directly. Recall checks the *context* against the reference; faithfulness checks the *answer* against the context. An answer can be faithful, relevant and still leave out half of what was found.
- Precision counts **blocks**, and the methods' blocks differ in size: OKF blocks are whole sections (up to 4,500 characters), graph blocks are chunks plus one block of facts. More, smaller blocks give more chances to rank useless ones high.

**Unanswerable questions** (the facts aren't in the papers) skip DeepEval. They're graded by one rule: did the method refuse?

## 4. The judge

DeepEval needs a model behind every metric. For OpenAI it has one built in. For Ollama we write a small wrapper (`DeepEvalBaseLLM` needs `load_model`, `generate`, `a_generate`, `get_model_name`) so we control three things:

- **`temperature: 0`**: the same answer gets the same verdict every time you run this.
- **`format` = the JSON schema DeepEval expects.** DeepEval asks for answers as JSON (lists of claims, yes/no verdicts). Small models often break JSON; Ollama's structured output forces valid JSON in the right shape. If it still fails to parse, the raw text goes back to DeepEval, which tries to repair it.
- **`num_ctx`**: Ollama's default context window would cut off the long contexts. `num_predict` caps each reply.

**The Mercury judge** is the same idea, talking to Inception's OpenAI-style API with `requests` (no extra package):
- **`response_format` = the JSON schema DeepEval expects**, like Ollama's `format`. If the API won't take a schema, the judge drops to plain JSON mode once and stays there (it prints when that happens).
- **`reasoning_effort`** from `MERCURY_REASONING`; any thinking text is stripped before the reply goes to DeepEval.
- **When the API says "slow down"** (HTTP 429) or is overloaded (5xx), or the Wi-Fi drops overnight, it waits (the time the API asks for, or longer each try) and tries again, up to about 15 minutes.
- **When the free tokens run out** (HTTP 402, or an error that mentions quota/credit), or the API stays down past those 15 minutes, it raises `StopScoring`. The scoring loop catches that, stops with everything so far saved, and tells you to run the cell again later.
- **If a reply isn't valid JSON** in the shape DeepEval wants, it asks again (up to 2 more times, the last one with more thinking and a doubled token cap) before handing the text to DeepEval's own repair.
- **Mercury's content filter** sometimes blocks a prompt (the Sparks of AGI penetration-testing passages, for example). The judge retries once with a short system message saying it is only grading text. If it's still blocked it raises `ContentFiltered`, and the loop saves that score as **skipped** instead of an error, so it isn't retried forever and doesn't count against the method.
- **It counts tokens** (prompt + completion) so you can see how much of the free budget a run used.

````python
class OllamaJudge(DeepEvalBaseLLM):
    # DeepEval judge backed by a local Ollama model

    def __init__(self, model_name, url=OLLAMA_URL, num_ctx=JUDGE_NUM_CTX, max_tokens=JUDGE_MAX_TOKENS, timeout=900):
        self.model_name = model_name
        self.url = url.rstrip("/")
        self.num_ctx = num_ctx
        self.max_tokens = max_tokens
        self.timeout = timeout
        self.calls, self.seconds, self.json_fallbacks = 0, 0.0, 0
        super().__init__(model_name)

    def load_model(self):
        return self.model_name

    def get_model_name(self):
        return f"ollama/{self.model_name}"

    def _chat(self, prompt, schema=None):
        body = {"model": self.model_name, "stream": False, "keep_alive": "30m",
                "messages": [{"role": "user", "content": prompt}],
                "options": {"temperature": 0, "num_ctx": self.num_ctx, "num_predict": self.max_tokens}}
        if schema is not None:
            body["format"] = schema.model_json_schema()
        start = time.perf_counter()
        resp = requests.post(f"{self.url}/api/chat", json=body, timeout=self.timeout)
        if resp.status_code >= 400 and schema is not None:     # older Ollama without JSON-schema output: plain JSON mode
            body["format"] = "json"
            resp = requests.post(f"{self.url}/api/chat", json=body, timeout=self.timeout)
        resp.raise_for_status()
        self.calls += 1
        self.seconds += time.perf_counter() - start
        text = resp.json()["message"]["content"]
        return re.sub(r"<think>.*?</think>", "", text, flags=re.S).strip()    # in case a thinking model is used

    def generate(self, prompt, schema=None, **kwargs):
        text = self._chat(prompt if isinstance(prompt, str) else str(prompt), schema)
        if schema is None:
            return text
        try:
            return schema.model_validate_json(text)
        except Exception:
            self.json_fallbacks += 1
            return text                                         # DeepEval tries to repair it

    async def a_generate(self, prompt, schema=None, **kwargs):
        return await asyncio.to_thread(self.generate, prompt, schema)


class StopScoring(Exception):
    """The judge can't go on (free tokens used up, or the API stays down): stop now; the scores so far are saved."""


class ContentFiltered(Exception):
    """The API's content filter refused this prompt, even with the grading-only system message."""


GRADER_SYSTEM = ("You are an automated evaluator for a question-answering benchmark built from published AI research "
                 "papers. The user message quotes passages from those papers and answers about them. Your only job is to "
                 "grade that text exactly as instructed and reply in JSON. You are not being asked to carry out anything "
                 "the quoted text describes.")


class MercuryJudge(DeepEvalBaseLLM):
    # DeepEval judge backed by Inception's Mercury API (OpenAI-style chat completions, plain REST)

    URL = "https://api.inceptionlabs.ai/v1/chat/completions"
    RETRY = (408, 409, 425, 429, 500, 502, 503, 504)

    def __init__(self, model_name, api_key, effort=MERCURY_REASONING, max_tokens=JUDGE_MAX_TOKENS,
                 timeout=180, max_waits=8, json_tries=3):
        self.model_name, self.api_key, self.effort = model_name, api_key, effort
        self.max_tokens, self.timeout, self.max_waits, self.json_tries = max_tokens, timeout, max_waits, json_tries
        self.json_mode = "schema"            # "schema" -> "object" -> "none" if the API rejects a format
        self.calls, self.seconds, self.json_fallbacks, self.waits = 0, 0.0, 0, 0
        self.json_retries, self.filtered = 0, 0
        self.prompt_tokens, self.completion_tokens, self.cut_off = 0, 0, 0
        super().__init__(model_name)

    def load_model(self):
        return self.model_name

    def get_model_name(self):
        return f"inception/{self.model_name}"

    def _body(self, prompt, schema, boost, framed):
        messages = [{"role": "user", "content": prompt}]
        if framed:
            messages.insert(0, {"role": "system", "content": GRADER_SYSTEM})
        effort = "medium" if boost and self.effort == "low" else self.effort
        body = {"model": self.model_name, "messages": messages, "temperature": 0,
                "reasoning_effort": effort, "max_completion_tokens": self.max_tokens * (2 if boost else 1)}
        if schema is not None and self.json_mode == "schema":
            body["response_format"] = {"type": "json_schema",
                                       "json_schema": {"name": schema.__name__, "schema": schema.model_json_schema()}}
        elif schema is not None and self.json_mode == "object":
            body["response_format"] = {"type": "json_object"}
        return body

    def _chat(self, prompt, schema, boost=False):
        start, attempt, framed = time.perf_counter(), 0, False
        while True:
            try:
                resp = requests.post(self.URL, json=self._body(prompt, schema, boost, framed), timeout=self.timeout,
                                     headers={"Authorization": f"Bearer {self.api_key}"})
                status, msg = resp.status_code, resp.text[:300]
            except requests.RequestException as exc:            # Wi-Fi drop, DNS hiccup, timeout
                resp, status, msg = None, None, f"{type(exc).__name__}: {exc}"[:300]
            if status == 200:
                break
            low = msg.lower()
            if "content_filter" in low:                          # blocked by the safety filter
                if not framed:
                    framed = True                                # once more, with the grading-only system message
                    continue
                self.filtered += 1
                raise ContentFiltered(msg[:150])
            if status == 400 and schema is not None and self.json_mode != "none" and \
                    any(w in low for w in ("response_format", "schema", "json")):
                self.json_mode = {"schema": "object", "object": "none"}[self.json_mode]
                print(f"   mercury rejected the JSON format, switching to json_mode={self.json_mode!r}: {msg[:120]}")
                continue
            if status in (401, 403):
                raise RuntimeError(f"Mercury {status}: check INCEPTION_API_KEY in .env ({msg[:120]})")
            if status == 402 or any(w in low for w in ("quota", "credit", "insufficient", "billing", "balance")):
                raise StopScoring(f"out of tokens / credit ({status}: {msg[:150]})")
            if (status is None or status in self.RETRY) and attempt < self.max_waits:
                asked = resp.headers.get("retry-after") if resp is not None else None
                delay = int(float(asked)) + 1 if asked and asked.replace(".", "").isdigit() else min(5 * 2 ** attempt, 300)
                print(f"   mercury {status or 'network error'}: waiting {delay} s ({msg[:80]})")
                self.waits += 1
                attempt += 1
                time.sleep(delay)
                continue
            if status is None or status in self.RETRY:
                raise StopScoring(f"Mercury still failing after {self.max_waits} waits: {status} {msg[:150]}")
            raise RuntimeError(f"Mercury error {status}: {msg[:200]}")   # saved as this score's error, retried next run
        data = resp.json()
        self.calls += 1
        self.seconds += time.perf_counter() - start
        usage = data.get("usage") or {}
        self.prompt_tokens += usage.get("prompt_tokens", 0)
        self.completion_tokens += usage.get("completion_tokens", 0)
        choice = (data.get("choices") or [{}])[0]
        if choice.get("finish_reason") == "length":
            self.cut_off += 1
        text = (choice.get("message") or {}).get("content") or ""
        return re.sub(r"<think>.*?</think>", "", text, flags=re.S).strip()

    @staticmethod
    def _parse(text, schema):
        for candidate in (text, (re.search(r"\{.*\}", text, re.S) or [text])[0]):   # whole reply, then the {...} inside it
            try:
                return schema.model_validate_json(candidate)
            except Exception:
                pass
        return None

    def generate(self, prompt, schema=None, **kwargs):
        prompt = prompt if isinstance(prompt, str) else str(prompt)
        if schema is None:
            return self._chat(prompt, None)
        for attempt in range(self.json_tries):            # last try: more thinking and a doubled token cap
            text = self._chat(prompt, schema, boost=attempt == self.json_tries - 1 and attempt > 0)
            parsed = self._parse(text, schema)
            if parsed is not None:
                return parsed
            if attempt < self.json_tries - 1:
                self.json_retries += 1
        self.json_fallbacks += 1
        return text                                             # DeepEval tries to repair it

    async def a_generate(self, prompt, schema=None, **kwargs):
        return await asyncio.to_thread(self.generate, prompt, schema)


if JUDGE_PROVIDER == "ollama":
    judge = OllamaJudge(JUDGE_MODEL)
    JUDGE_NAME = judge.get_model_name()
elif JUDGE_PROVIDER == "mercury":
    assert INCEPTION_API_KEY, "no Inception key: put INCEPTION_API_KEY=... in .env"
    judge = MercuryJudge(JUDGE_MODEL, INCEPTION_API_KEY)
    JUDGE_NAME = judge.get_model_name()
elif JUDGE_PROVIDER == "openai":
    assert os.getenv("OPENAI_API_KEY"), "put OPENAI_API_KEY in .env"
    judge = JUDGE_MODEL                     # DeepEval's built-in OpenAI judge
    JUDGE_NAME = f"openai/{JUDGE_MODEL}"
else:
    raise ValueError(JUDGE_PROVIDER)

# smoke test: one structured call, shaped like DeepEval's faithfulness verdicts
if JUDGE_PROVIDER in ("ollama", "mercury"):
    from typing import List, Optional
    from pydantic import BaseModel

    class Verdict(BaseModel):
        verdict: str
        reason: Optional[str] = None

    class Verdicts(BaseModel):
        verdicts: List[Verdict]

    t0 = time.perf_counter()
    print(judge.generate('Context: "Llama 2 was pretrained on 2 trillion tokens."\n'
                         'Claims: ["Llama 2 was pretrained on 2T tokens.", "Llama 2 has 80 layers."]\n'
                         'For each claim, in order, give a verdict: "yes" (the context supports it), "no" (it contradicts it) '
                         'or "idk" (the context does not say), plus a short reason. '
                         'Reply as JSON: {"verdicts": [{"verdict": ..., "reason": ...}]}', schema=Verdicts))
    print(f"{time.perf_counter() - t0:.1f} s (expected: yes, then idk)")
    if JUDGE_PROVIDER == "mercury":
        print(f"json_mode: {judge.json_mode} | tokens so far: {judge.prompt_tokens} in, {judge.completion_tokens} out")
````

## 5. Refusals

All four methods are told to say "The bundle / graph / knowledge **does not contain this.**" when the context doesn't have the answer. Some answers word it differently ("The knowledge does not contain the exact score…"), so the check looks for "does not contain" after any of those words, not only "…contain this". That gives two checks with no judge at all:

- **unanswerable questions**: refusing is the right answer.
- **answerable questions**: a refusal means the method missed something it should have found.

A refusal counts as **full** when the answer is only that sentence, and **partial** when it comes with other text (common on two-part questions, where one half is answered).

A full refusal has no claims, so DeepEval would give it a perfect faithfulness of 1.0. To keep that from flattering a method, **faithfulness is skipped for full refusals**. Answer relevancy still scores them (low), and the retrieval metrics don't look at the answer at all.

````python
REFUSAL_RE = re.compile(r"\b(bundle|graph|knowledge|context)\s+(does not|doesn't|do not)\s+contain(\s+this)?\b", re.I)


def refusal(answer):
    if not REFUSAL_RE.search(answer):
        return None
    rest = re.sub(r"\[[^\]]*\]|[\s.,;:]", "", REFUSAL_RE.sub("", answer))   # drop the sentence, citations, punctuation
    return "full" if len(rest) < 40 else "partial"


rows = []
for q in questions:
    for method, label in METHODS.items():
        r = result_for(method, q)
        if r:
            rows.append({"method": label, "answerable": q["answerable"], "refusal": refusal(r["answer"])})
ref = pd.DataFrame(rows)
table = pd.DataFrame({
    "answerable: full refusals": ref[ref.answerable].groupby("method").refusal.apply(lambda s: f"{(s == 'full').sum()}/{len(s)}"),
    "answerable: partial": ref[ref.answerable].groupby("method").refusal.apply(lambda s: f"{(s == 'partial').sum()}/{len(s)}"),
    "unanswerable: refused (good)": ref[~ref.answerable].groupby("method").refusal.apply(lambda s: f"{s.notna().sum()}/{len(s)}"),
}).reindex([l for l in METHODS.values() if l in set(ref.method)])
print(table.to_string())
````

## 6. Score everything

For every answerable question and every method, this builds a DeepEval `LLMTestCase` from the files and runs the four metrics:

```python
LLMTestCase(input=question, actual_output=answer, expected_output=reference_answer, retrieval_context=blocks)
```

How the loop is set up:
- **Question by question, all four methods each time.** If you stop it halfway, the methods have been graded on the same questions, so the tables below still compare like with like. They only use questions that every method has scores for.
- **Checkpointed.** Every score is appended to `artifacts/results/scores.jsonl` as soon as it exists. Re-running skips what's done, so you can stop and continue later.
- **Fingerprinted.** Each score remembers a hash of the question, reference, answer and context. Re-run a method's notebook and its changed answers get re-graded; unchanged ones don't.
- **Errors are kept and retried.** If the judge returns something DeepEval can't parse, that score is saved as an error and tried again on the next run.

With Mercury: if the free tokens run out or the API stays down for ~15 minutes, the loop stops with a message, and the next run picks up from there. Each metric in the summary uses only the questions where all four methods have that metric, so a partial run is still a fair comparison.

Time: Mercury answers in a few seconds per call, and there are ~27 answerable questions × 4 methods × 6–10 calls (~750 calls). Expect well under an hour, a bit more if it has to wait out rate limits. Start with `QUESTION_LIMIT = 3` to check it, then set it back to `None` for the full run.

````python
METRICS = {
    "contextual_precision": ContextualPrecisionMetric,
    "contextual_recall": ContextualRecallMetric,
    "faithfulness": FaithfulnessMetric,
    "answer_relevancy": AnswerRelevancyMetric,
}
SHORT = {"contextual_precision": "prec", "contextual_recall": "rec", "faithfulness": "faith", "answer_relevancy": "rel"}


def fingerprint(q, r):
    blob = json.dumps([q["question"], q["expected_output"], r["answer"], r["retrieval_context"]], ensure_ascii=False)
    return hashlib.sha1(blob.encode("utf-8")).hexdigest()[:12]


def load_scores():
    scores = {}
    if SCORES_PATH.exists():
        for line in SCORES_PATH.open(encoding="utf-8"):
            if line.strip():
                s = json.loads(line)
                scores[(s["judge"], s["method"], s["id"], s["metric"])] = s     # later lines win
    return scores


def measure(name, q, r):
    test_case = LLMTestCase(input=q["question"], actual_output=r["answer"], expected_output=q["expected_output"],
                            retrieval_context=r["retrieval_context"])
    metric = METRICS[name](threshold=THRESHOLD, model=judge, include_reason=INCLUDE_REASON,
                           async_mode=False, verbose_mode=False)
    metric.measure(test_case, _show_indicator=False)
    return float(metric.score), metric.reason


def fmt(x):
    return " -- " if x is None else f"{x:.2f}"
````

````python
scores = load_scores()
todo = [q for q in questions if q["answerable"]][:QUESTION_LIMIT]
calls_before = getattr(judge, "calls", 0)
run_start, n_new, stopped = time.perf_counter(), 0, None


def score_one(n, q, method, label, out):
    global n_new
    r = result_for(method, q)
    if r is None:
        return
    fp, full_refusal = fingerprint(q, r), refusal(r["answer"]) == "full"
    line, t0 = [], time.perf_counter()
    for name in METRICS:
        key = (JUDGE_NAME, method, q["id"], name)
        old = scores.get(key)
        if old and old["fingerprint"] == fp and old["error"] is None:
            line.append(f"{SHORT[name]} {fmt(old['score'])}")
            continue
        rec = {"judge": JUDGE_NAME, "method": method, "id": q["id"], "metric": name, "fingerprint": fp,
               "score": None, "reason": None, "error": None, "skipped": None, "seconds": 0.0}
        if name == "faithfulness" and full_refusal:
            rec["skipped"] = "full refusal"
        else:
            t1 = time.perf_counter()
            try:
                rec["score"], rec["reason"] = measure(name, q, r)
            except StopScoring:
                raise                                   # out of tokens / API down: stop everything (nothing half-written)
            except ContentFiltered:
                rec["skipped"] = "content filter"       # saved, not retried; left out of that metric for every method
            except Exception as e:
                rec["error"] = f"{type(e).__name__}: {e}"[:300]
            rec["seconds"] = round(time.perf_counter() - t1, 1)
            n_new += 1
        out.write(json.dumps(rec, ensure_ascii=False) + "\n")
        out.flush()
        scores[key] = rec
        line.append(f"{SHORT[name]} {'ERR ' if rec['error'] else 'FILT' if rec['skipped'] == 'content filter' else fmt(rec['score'])}")
    print(f"[{n:2}/{len(todo)}] {q['id']:4} {label:12} " + " | ".join(line) + f"   ({time.perf_counter() - t0:.0f} s)")


with SCORES_PATH.open("a", encoding="utf-8") as out:
    try:
        for n, q in enumerate(todo, start=1):
            for method, label in METHODS.items():
                score_one(n, q, method, label, out)
    except StopScoring as exc:
        stopped = str(exc)

elapsed = time.perf_counter() - run_start
print(f"\n{n_new} new scores in {elapsed / 60:.1f} min", end="")
if hasattr(judge, "calls"):
    calls = judge.calls - calls_before
    print(f" | {calls} judge calls, {judge.seconds / max(judge.calls, 1):.1f} s each, "
          f"{judge.json_fallbacks} JSON repairs", end="")
if hasattr(judge, "prompt_tokens"):
    print(f"\ntokens this session: {judge.prompt_tokens:,} in + {judge.completion_tokens:,} out "
          f"= {judge.prompt_tokens + judge.completion_tokens:,} | {judge.waits} waits | "
          f"{judge.cut_off} replies cut off | {judge.json_retries} JSON retries | {judge.filtered} content-filtered | "
          f"json_mode {judge.json_mode!r}")
errors = sum(1 for s in scores.values() if s["judge"] == JUDGE_NAME and s["error"])
print(f" | {errors} errors (re-run this cell to retry them)")
if stopped:
    print(f"\nSTOPPED: {stopped}. Everything above is saved; run this cell again later to continue.")
````

## 7. Results

One row per method. Each metric is averaged over the answerable questions where **every** method has a score for that metric (so one failed score drops that question from one metric, not from all four). The `n` row says how many questions each metric used. The other columns:

- **answered %**: answerable questions that didn't get a full refusal.
- **unanswerable refused**: how many unanswerable questions were (correctly) refused.
- **retrieval / generation ms**: medians from the results files. These come from your machine, so they're comparable only because all four notebooks ran on it with the same model.
- **prompt / answer tokens**: means. Prompt tokens is the cost of the context.

Small differences are noise: with ~27 questions, the ± table under the summary shows how far each average could move by chance. If two methods' ranges overlap a lot, don't call a winner on that metric.

````python
scores = load_scores()
rows = []
for q in questions:
    for method, label in METHODS.items():
        r = result_for(method, q)
        if r is None:
            continue
        fp = fingerprint(q, r)
        row = {"id": q["id"], "type": q["type"], "papers": q["papers"], "answerable": q["answerable"], "method": label,
               "refusal": refusal(r["answer"]), "retrieval_ms": r["retrieval_ms"], "generation_ms": r["generation_ms"],
               "prompt_tokens": r["prompt_tokens"], "answer_tokens": r["answer_tokens"],
               "context_blocks": len(r["retrieval_context"]), "context_chars": sum(map(len, r["retrieval_context"]))}
        for name in METRICS:
            s = scores.get((JUDGE_NAME, method, q["id"], name))
            ok = s is not None and s["fingerprint"] == fp and s["error"] is None and s.get("skipped") != "content filter"
            row[name] = s["score"] if ok else None
            row[name + "_done"] = ok
        rows.append(row)
df = pd.DataFrame(rows)
METRIC_COLS = list(METRICS)
LABELS = [l for l in METHODS.values() if l in set(df.method)]

# per metric: the answerable questions where all four methods have that metric
ans = df[df.answerable]
per_method = ans.groupby("id").method.nunique()
common = {m: [i for i, ok in ans.groupby("id")[m + "_done"].all().items() if ok and per_method[i] == len(METHODS)]
          for m in METRIC_COLS}
graded = ans[ans.id.isin(set().union(*common.values()))].copy()
for m in METRIC_COLS:
    graded.loc[~graded.id.isin(common[m]), m] = None     # this question doesn't count for this metric
unans = df[~df.answerable]
print(f"graded by {JUDGE_NAME}: questions used per metric (of {len(per_method)} answerable): "
      + ", ".join(f"{SHORT[m]} {len(common[m])}" for m in METRIC_COLS))

summary = pd.DataFrame({
    **{m: graded.groupby("method")[m].mean() for m in METRIC_COLS},
    "answered %": df[df.answerable].groupby("method").refusal.apply(lambda s: 100 * (s != "full").mean()),
    "unanswerable refused": unans.groupby("method").refusal.apply(lambda s: f"{s.notna().sum()}/{len(s)}"),
    "retrieval ms": df.groupby("method").retrieval_ms.median(),
    "generation ms": df.groupby("method").generation_ms.median(),
    "prompt tokens": df.groupby("method").prompt_tokens.mean(),
    "answer tokens": df.groupby("method").answer_tokens.mean(),
}).reindex(LABELS)
summary.loc["n"] = pd.Series({m: len(common[m]) for m in METRIC_COLS})

margin = pd.DataFrame({m: 1.96 * graded.groupby("method")[m].std() / graded.groupby("method")[m].count() ** 0.5
                       for m in METRIC_COLS}).reindex(LABELS)

try:
    display(summary.style.format(precision=3, na_rep="").format(precision=0, subset=pd.IndexSlice[["n"], METRIC_COLS]).format(precision=0, subset=["answered %", "retrieval ms", "generation ms",
                                                                          "prompt tokens", "answer tokens"])
            .highlight_max(subset=pd.IndexSlice[LABELS, METRIC_COLS], color="#d4edda"))
except Exception:
    display(summary.round(3))
print("± (95% range of each average):")
display(margin.round(3))
````

````python
def show_question(qid, answer_chars=700):
    q = by_id[qid]
    print(f"{qid} [{q['type']}] {q['question']}\n")
    print("REFERENCE:", q["expected_output"], "\n")
    for method, label in METHODS.items():
        r = result_for(method, q)
        if r is None:
            print(f"=== {label}: no result\n")
            continue
        fp = fingerprint(q, r)
        line = []
        for name in METRICS:
            s = scores.get((JUDGE_NAME, method, qid, name))
            ok = s is not None and s["fingerprint"] == fp
            line.append(f"{SHORT[name]} {fmt(s['score']) if ok else ' -- '}")
        print(f"=== {label}: " + " | ".join(line) + f" | refusal: {refusal(r['answer'])}")
        print(r["answer"][:answer_chars] + (" ..." if len(r["answer"]) > answer_chars else ""))
        print("context:", " / ".join(block.strip().splitlines()[0][:60] for block in r["retrieval_context"]))
        if INCLUDE_REASON:
            for name in METRICS:
                s = scores.get((JUDGE_NAME, method, qid, name))
                if s and s.get("reason"):
                    print(f"  {SHORT[name]}: {s['reason']}")
        print()


show_question(questions[0]["id"])
````

### Judge vs judge

`scores.jsonl` keeps every judge's scores, so you can see how much the judge itself changes the numbers. This compares the current judge with each earlier one, using only the answers both graded. A big gap on a metric means that metric depends more on who is judging than on the method.

````python
every = pd.DataFrame(load_scores().values())
every = every[every.error.isna() & every.score.notna()]
others = [j for j in sorted(every.judge.unique()) if j != JUDGE_NAME]
if not others:
    print("only one judge so far:", JUDGE_NAME)
for other in others:
    pair = every[every.judge.isin([JUDGE_NAME, other])]
    both = pair.pivot_table(index=["method", "id", "metric", "fingerprint"], columns="judge", values="score").dropna()
    print(f"\n{JUDGE_NAME} vs {other}: {len(both)} scores graded by both")
    if both.empty:
        continue
    by_metric = both.groupby(level="metric").mean()
    by_metric["mean |difference|"] = (both[JUDGE_NAME] - both[other]).abs().groupby(level="metric").mean()
    display(by_metric.round(2))
    display(both.groupby(level="method").mean().rename(index={m: METHODS[m] for m in METHODS}).round(2))
````

## How far to trust these numbers

- **The judge is a model too.** A 4B judge misses subtle errors and can be inconsistent on long contexts, and when it grades its own answers it tends to be generous. Mercury is a different family, but it's a speed-first model, so check it on a few answers you already know are right or wrong. Before trusting a gap between two methods, read a few of the questions where they differ with `show_question`, and if it matters, re-run with a stronger judge. The checkpoint keeps each judge's scores separately.
- **~27 answerable questions is a small sample.** Look at the ± table. Per-type tables have 3–8 questions per cell, so read them as hints.
- **The questions were written from random sections, not from the OKF topic pages**, so the OKF methods don't get questions shaped like their own summaries. The reference answers were checked against the chunk text and the bundle, but still read them: if you disagree with one, fix it in `eval_questions.jsonl` and re-run (only the changed question gets re-graded).
- **Block size differs between methods** (section vs chunk), which affects contextual precision more than the other three.
- **Missing: answer correctness.** None of the four compares the answer with the reference. If you want that, DeepEval's `GEval` with a "does the answer contain the reference's key facts?" criterion adds it as a fifth metric.
